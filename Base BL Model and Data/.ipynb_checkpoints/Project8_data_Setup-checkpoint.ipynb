```python
import yfinance as yf
import pandas as pd

# Define tickers
tickers = {
    "EURUSD": "EURUSD=X",
    "GBPUSD": "GBPUSD=X",
    "GOLD":   "GC=F",
    "OIL":    "CL=F",
}

start_date = "2018-01-01"
end_date   = "2026-06-30"

all_data = []

for name, ticker in tickers.items():
    print(f"Downloading {name}...")
    df = yf.download(ticker, start=start_date,
                     end=end_date, auto_adjust=True)

    # Keep all 5 columns
    ohlcv = df[["Open","High","Low","Close","Volume"]].copy()

    # Rename each column to include asset name
    # e.g. EURUSD_Open, EURUSD_High, etc.
    ohlcv.columns = [
        f"{name}_Open",
        f"{name}_High",
        f"{name}_Low",
        f"{name}_Close",
        f"{name}_Volume",
    ]

    all_data.append(ohlcv)
    print(f"  Got {len(ohlcv)} rows for {name}")

# Combine all assets
market_df = pd.concat(all_data, axis=1)
market_df.index.name = "Date"

# Remove weekends
market_df = market_df[market_df.index.dayofweek < 5]

# Save
market_df.to_csv("market_data.csv")
print("\nDone! Saved to market_data.csv")
print(f"Total rows: {len(market_df)}")
print(f"Total columns: {len(market_df.columns)}")
print("\nColumn names:")
print(list(market_df.columns))
print("\nLast 3 rows:")
print(market_df.tail(3))
```

```python
"""
Project 8 — build the combined market + macro + policy dataset  (robust to either policy schema).

Inputs:
  - market_data.csv   (OHLCV: EURUSD, GBPUSD, GOLD, OIL)          <- from the notebook
  - macro_data.csv    (FED_RATE, INFLATION, USD_INDEX, VIX, US10Y_YIELD)
  - policy_events.csv  (NEW 6-col: Date,Event_Type,Description,Severity,Assets_Affected,Period
                        OR OLD 5-col: date,type,severity,description,source)

Output: project8_dataset_v2.csv  (33-column format).
Falls back to reusing market+macro columns from project8_dataset.csv if the raw files are absent.
"""
import os
import pandas as pd

SPLIT_DATE = pd.Timestamp("2025-01-01")   # TRAIN < this <= TEST
BASE_MARKET_MACRO = [
    "EURUSD_Open","EURUSD_High","EURUSD_Low","EURUSD_Close","EURUSD_Volume",
    "GBPUSD_Open","GBPUSD_High","GBPUSD_Low","GBPUSD_Close","GBPUSD_Volume",
    "GOLD_Open","GOLD_High","GOLD_Low","GOLD_Close","GOLD_Volume",
    "OIL_Open","OIL_High","OIL_Low","OIL_Close","OIL_Volume",
    "FED_RATE","INFLATION","USD_INDEX","VIX","US10Y_YIELD",
]
SEV_RANK = {"LOW":1, "MEDIUM":2, "HIGH":3, "SEVERE":4}
ASSET_ORDER = ["EURUSD","GBPUSD","GOLD","OIL"]
# type -> default affected assets, used ONLY if the policy file has no assets column
ASSET_MAP = {
    "TARIFF":"EURUSD,GBPUSD,GOLD,OIL", "TRADE":"EURUSD,GBPUSD,GOLD,OIL",
    "RATE":"EURUSD,GBPUSD,GOLD", "OIL_SUPPLY":"OIL", "SANCTIONS":"EURUSD,GOLD,OIL",
    "COURT_RULING":"EURUSD,GBPUSD,GOLD,OIL", "GEOPOLITICAL":"GOLD,OIL",
    "CURRENCY":"EURUSD,GBPUSD,GOLD", "BANKING":"GOLD,EURUSD", "SHOCK":"EURUSD,GBPUSD,GOLD,OIL",
    "POLICY":"EURUSD,GBPUSD,GOLD,OIL",
}

# ---- 1. market + macro base ------------------------------------------------
if os.path.exists("market_data.csv") and os.path.exists("macro_data.csv"):
    market = pd.read_csv("market_data.csv", parse_dates=["Date"])
    macro  = pd.read_csv("macro_data.csv",  parse_dates=["Date"])
    base = market.merge(macro, on="Date", how="inner").sort_values("Date")
    print("Base built from market_data.csv + macro_data.csv")
else:
    src = pd.read_csv("project8_dataset.csv", parse_dates=["Date"])
    base = src[["Date"] + BASE_MARKET_MACRO].copy().sort_values("Date")
    print("market_data/macro_data not found -> reusing market+macro from project8_dataset.csv")
base = base.reset_index(drop=True)
trading_days = base["Date"].sort_values().reset_index(drop=True)

# ---- 2. policy events (accepts EITHER schema) ------------------------------
ev = pd.read_csv("policy_events.csv")                 # read WITHOUT parse_dates first
ev.columns = [c.strip() for c in ev.columns]

# normalise column names from whichever schema was provided
rename_any = {
    "date":"Date", "Date":"Date",
    "type":"POLICY_TYPE", "Event_Type":"POLICY_TYPE",
    "description":"POLICY_DESCRIPTION", "Description":"POLICY_DESCRIPTION",
    "severity":"POLICY_SEVERITY", "Severity":"POLICY_SEVERITY",
    "assets":"POLICY_ASSETS", "Assets_Affected":"POLICY_ASSETS", "assets_affected":"POLICY_ASSETS",
    "period":"POLICY_PERIOD", "Period":"POLICY_PERIOD",
}
ev = ev.rename(columns={c: rename_any[c] for c in ev.columns if c in rename_any})

assert "Date" in ev.columns, f"no date column found in policy_events.csv (got {list(ev.columns)})"
ev["Date"] = pd.to_datetime(ev["Date"])
ev["POLICY_TYPE"] = ev["POLICY_TYPE"].astype(str).str.upper()
ev["POLICY_SEVERITY"] = ev["POLICY_SEVERITY"].astype(str).str.upper()

# fill anything the file didn't supply
if "POLICY_DESCRIPTION" not in ev.columns:
    ev["POLICY_DESCRIPTION"] = ""
if "POLICY_ASSETS" not in ev.columns:
    ev["POLICY_ASSETS"] = ev["POLICY_TYPE"].map(lambda t: ASSET_MAP.get(t, "EURUSD,GBPUSD,GOLD,OIL"))
if "POLICY_PERIOD" not in ev.columns:
    ev["POLICY_PERIOD"] = ev["Date"].map(lambda d: "TRAIN" if d < SPLIT_DATE else "TEST")

# ---- 3. snap each event forward to the next trading day --------------------
def snap_forward(d):
    i = trading_days.searchsorted(d, side="left")
    return trading_days.iloc[i] if i < len(trading_days) else pd.NaT
ev["mapped_date"] = ev["Date"].map(snap_forward)
n_direct  = int((ev["Date"] == ev["mapped_date"]).sum())
n_snapped = int((ev["Date"] != ev["mapped_date"]).sum())
ev = ev.dropna(subset=["mapped_date"])

# ---- 4. merge events landing on the same trading day ----------------------
def merge_assets(series):
    have = set()
    for a in series: have.update(str(a).split(","))
    return ",".join([a for a in ASSET_ORDER if a in have])

agg = ev.groupby("mapped_date").agg(
    POLICY_TYPE=("POLICY_TYPE", lambda s: "/".join(sorted(set(s)))),
    POLICY_SEVERITY=("POLICY_SEVERITY", lambda s: max(s, key=lambda x: SEV_RANK.get(str(x).upper(),0))),
    POLICY_DESCRIPTION=("POLICY_DESCRIPTION", lambda s: " | ".join(map(str, s))),
    POLICY_ASSETS=("POLICY_ASSETS", merge_assets),
    POLICY_PERIOD=("POLICY_PERIOD", lambda s: s.iloc[0]),
).reset_index().rename(columns={"mapped_date":"Date"})
n_collisions = len(ev) - len(agg)

# ---- 5. assemble the final 33-column schema -------------------------------
out = base.merge(agg, on="Date", how="left")
out["POLICY_EVENT"] = out["POLICY_TYPE"].notna().astype(int)
out["PERIOD"]       = out["Date"].map(lambda d: "TRAIN" if d < SPLIT_DATE else "TEST")
out.loc[out["POLICY_EVENT"] == 0, "POLICY_PERIOD"] = pd.NA

COLS = ["Date"] + BASE_MARKET_MACRO + [
    "POLICY_EVENT","POLICY_TYPE","POLICY_DESCRIPTION","POLICY_SEVERITY",
    "POLICY_ASSETS","POLICY_PERIOD","PERIOD"]
out = out[COLS]
out.to_csv("project8_dataset_v2.csv", index=False)

# ---- 6. report -------------------------------------------------------------
print("\n=== SUMMARY ===")
print(f"rows: {len(out)}  cols: {out.shape[1]}  date: {out.Date.min().date()} -> {out.Date.max().date()}")
print(f"policy events: {len(ev)}  (direct {n_direct}, snapped {n_snapped}, merged {n_collisions})")
print(f"event rows flagged: {int(out.POLICY_EVENT.sum())}  "
      f"(TRAIN {int(((out.POLICY_EVENT==1)&(out.PERIOD=='TRAIN')).sum())}, "
      f"TEST {int(((out.POLICY_EVENT==1)&(out.PERIOD=='TEST')).sum())})")
print("types:", sorted(out.POLICY_TYPE.dropna().unique()))
```
