```python
import pandas as pd
import numpy as np
from pypfopt import risk_models, EfficientFrontier
from pypfopt.black_litterman import BlackLittermanModel

df = pd.read_csv("project8_dataset_v2.csv", parse_dates=["Date"], index_col="Date")

assets = ["EURUSD", "GBPUSD", "GOLD", "OIL"]
prices = df[[f"{a}_Close" for a in assets]].copy()
prices.columns = assets
prices = prices.dropna()

print("Prices shape:", prices.shape)
print(prices.tail(3).round(4))
```

```python
# WTI oil went negative on 2020-04-20 (-$37.63). pct_change through
# zero/negative prices gives meaningless returns (-306%, +127%) that
# inflate the covariance ~10x. Standard treatment: winsorize.

returns = prices.pct_change().dropna()

CLIP = 0.20  # cap daily returns at +/-20%
returns_clean = returns.clip(lower=-CLIP, upper=CLIP)

n_clipped = (returns.abs() > CLIP).sum()
print("Days clipped per asset:")
print(n_clipped)
print("\nAnnualized vol before / after fix:")
print(pd.DataFrame({
    "before": returns.std() * np.sqrt(252),
    "after":  returns_clean.std() * np.sqrt(252),
}).round(3))
```

```python
from pypfopt import risk_models

train_dates = df[df["PERIOD"] == "TRAIN"].index
prices_train = prices.loc[prices.index.isin(train_dates)]
# --- Data fix: April 2020 negative oil price episode ---
# WTI futures went negative on 2020-04-20 (-$37.63); pct_change
# through zero produces meaningless returns that inflate the
# covariance matrix ~10x. Standard treatment: exclude the episode.

returns = prices.pct_change().dropna()

# Option A (recommended): winsorize — cap daily returns at ±20%
returns_clean = returns.clip(lower=-0.20, upper=0.20)

S = risk_models.CovarianceShrinkage(prices_train).ledoit_wolf()

delta = 2.5
w_eq = np.ones(len(assets)) / len(assets)
pi = pd.Series(delta * S.values @ w_eq, index=assets)
print(pi.round(4))
```

```python
train_dates = df[df["PERIOD"] == "TRAIN"].index
train_rets = returns_clean.loc[returns_clean.index.isin(train_dates)]
print("Train return rows:", len(train_rets))

# Ledoit-Wolf shrinkage on CLEAN RETURNS (returns_data=True is the key flag)
S = risk_models.CovarianceShrinkage(train_rets, returns_data=True).ledoit_wolf()
print("\nAnnualized covariance S:")
print(S.round(4))

# Equal-weight implied prior:  pi = delta * S * w_eq
delta = 2.5
w_eq = np.ones(len(assets)) / len(assets)
pi = pd.Series(delta * S.values @ w_eq, index=assets)
print("\nPrior (equilibrium) returns:")
print(pi.round(4))
```

```python
bl_base = BlackLittermanModel(S, pi=pi, absolute_views={})
ret_base = bl_base.bl_returns()
cov_base = bl_base.bl_cov()

ef = EfficientFrontier(ret_base, cov_base, weight_bounds=(0, 1))
ef.max_quadratic_utility(risk_aversion=delta)
w_base = pd.Series(ef.clean_weights())
print("Baseline weights (no views):")
print(w_base.round(3))
```

```python
# Simulating what the policy-shock / LLM layers will inject later
views = {"OIL": -0.08, "GOLD": 0.06}
confidences = [0.8, 0.6]   # same order as the views dict

bl_v = BlackLittermanModel(S, pi=pi,
                           absolute_views=views,
                           view_confidences=confidences,
                           omega="idzorek")
ret_v = bl_v.bl_returns()
cov_v = bl_v.bl_cov()

ef2 = EfficientFrontier(ret_v, cov_v, weight_bounds=(0, 1))
ef2.max_quadratic_utility(risk_aversion=delta)
w_v = pd.Series(ef2.clean_weights())

print("Prior vs posterior returns:")
print(pd.DataFrame({"prior": pi, "posterior": ret_v}).round(4))
print("\nBaseline vs with-views weights:")
print(pd.DataFrame({"baseline": w_base, "with_views": w_v}).round(3))
```

```python
def run_bl(views: dict = {}, confidences: list | None = None):
    """Member 1 interface: all other members call ONLY this.
    views: {"ASSET": expected_return}, confidences: [0-1] per view."""
    if views and confidences:
        bl = BlackLittermanModel(S, pi=pi, absolute_views=views,
                                 view_confidences=confidences,
                                 omega="idzorek")
    else:
        bl = BlackLittermanModel(S, pi=pi, absolute_views=views)
    ret, cov = bl.bl_returns(), bl.bl_cov()
    ef = EfficientFrontier(ret, cov, weight_bounds=(0, 1))
    ef.max_quadratic_utility(risk_aversion=delta)
    return ret, cov, pd.Series(ef.clean_weights())

# Sanity checks — run after every change
r, c, w = run_bl()
assert abs(w.sum() - 1) < 1e-6, "weights must sum to 1"
assert (w >= -1e-9).all(), "long-only violated"
r2, c2, w2 = run_bl({"OIL": -0.08}, [0.8])
assert r2["OIL"] < r["OIL"], "negative view must lower OIL posterior"
print("All sanity checks PASSED")
print("\nInterface ready:  run_bl(views, confidences) -> (returns, cov, weights)")
```

```python
sev_map = {"LOW": 0.3, "MEDIUM": 0.5, "HIGH": 0.7, "SEVERE": 0.9}

def event_to_confidence(row):
    return sev_map[row["POLICY_SEVERITY"]], row["POLICY_ASSETS"].split(",")

events = df[df["POLICY_EVENT"] == 1]
conf, affected = event_to_confidence(events.iloc[0])
print(f"Example event {events.index[0].date()}: "
      f"type={events.iloc[0]['POLICY_TYPE']}, "
      f"confidence={conf}, assets={affected}")
```

```python
# Monthly-rebalanced backtest over the TEST period (no look-ahead:
# each rebalance uses only data available up to that date)

test_dates = df[df["PERIOD"] == "TEST"].index
test_rets = returns_clean.loc[returns_clean.index.isin(test_dates)]

WINDOW = 252  # trailing 1-year estimation window

rebalance_dates = test_rets.resample("ME").last().index  # month-ends
portfolio_rets = []
weight_history = {}

current_w = np.ones(len(assets)) / len(assets)  # start equal-weight

for date in test_rets.index:
    # daily portfolio return with current weights
    portfolio_rets.append(current_w @ test_rets.loc[date].values)

    # month-end -> re-estimate and rebalance
    if date in rebalance_dates or date == test_rets.index[0]:
        hist = returns_clean.loc[:date].tail(WINDOW)
        S_t = risk_models.CovarianceShrinkage(hist, returns_data=True).ledoit_wolf()
        pi_t = pd.Series(delta * S_t.values @ w_eq, index=assets)
        bl_t = BlackLittermanModel(S_t, pi=pi_t, absolute_views={})
        ef_t = EfficientFrontier(bl_t.bl_returns(), bl_t.bl_cov(),
                                 weight_bounds=(0, 1))
        ef_t.max_quadratic_utility(risk_aversion=delta)
        current_w = np.array(list(ef_t.clean_weights().values()))
        weight_history[date] = current_w

bt = pd.Series(portfolio_rets, index=test_rets.index, name="BL_baseline")
```

```python
def metrics(r, name):
    ann_ret = (1 + r).prod() ** (252 / len(r)) - 1
    ann_vol = r.std() * np.sqrt(252)
    sharpe = ann_ret / ann_vol
    cum = (1 + r).cumprod()
    max_dd = (cum / cum.cummax() - 1).min()
    return pd.Series({"ann_return": ann_ret, "ann_vol": ann_vol,
                      "sharpe": sharpe, "max_drawdown": max_dd}, name=name)

# Comparison: equal-weight buy-and-hold
ew_rets = test_rets @ w_eq

results = pd.concat([
    metrics(bt, "BL_baseline"),
    metrics(ew_rets, "Equal_weight"),
], axis=1).round(4)
print(results)

(1 + pd.DataFrame({"BL": bt, "EW": ew_rets})).cumprod().plot(
    figsize=(10, 5), title="Cumulative return — TEST period")
```
