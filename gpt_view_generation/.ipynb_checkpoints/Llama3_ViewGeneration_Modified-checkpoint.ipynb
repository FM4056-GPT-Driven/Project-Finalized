Continues directly from your teammate's **Step 4B (FAISS Episodic Memory, `_v2`)** and
**Step 3 (Lag-Llama World Model, `lag_llama_views.csv`)**.

**What this notebook does:**
1. Loads Lag-Llama's per-asset forecast (Q) + confidence from Step 3

So this notebook's final output MUST be a dict of `{asset: expected_return}` plus a
matching list of confidences in [0, 1] — nothing fancier.

## Section 3 — Load Step 4B artifacts (FAISS index, episodes, scaler)

Loading the exact `_v2` files your teammate saved — 24-dim embeddings, not the original 84.
We also redefine `build_embedding`, `retrieve_similar_episodes`, `retrieve_similar_episodes_embargoed`,
and `format_episodes_for_llama` here so this notebook is self-contained and doesn't depend on
re-running the whole FAISS notebook first.


````python
WINDOW_SIZE = 20
FORECAST_HORIZON = 30
CLIP = 0.20
assets = ["EURUSD_Close", "GBPUSD_Close", "GOLD_Close", "OIL_Close"]
macro_cols = ["VIX", "USD_INDEX", "US10Y_YIELD"]

index = faiss.read_index("episodic_memory_v2.index")
with open("episodes_v2.pkl", "rb") as f:
    episodes = pickle.load(f)
with open("scaler_v2.pkl", "rb") as f:
    scaler = pickle.load(f)

print(f"Loaded {index.ntotal} episodes | embedding dim: {index.d}")
assert index.d == 24, "expected the corrected 24-dim embeddings from Step 4B v2"


def build_embedding(episode):
    price = episode["price_returns"].reshape(WINDOW_SIZE, len(assets))
    price = np.clip(price, -CLIP, CLIP)
    price_summary = []
    for a in range(len(assets)):
        col = price[:, a]
        price_summary += [col.mean(), col.std(), col.sum(), col.max(), col.min()]
    macro_vals = episode["macro_features"]
    policy = [episode["policy_event"]]
    return np.array(price_summary + list(macro_vals) + policy)


def retrieve_similar_episodes(current_episode, k=5):
    query_vec = build_embedding(current_episode)
    query_vec = scaler.transform([query_vec]).astype("float32")
    distances, idxs = index.search(query_vec, k)
    results = []
    for dist, idx in zip(distances[0], idxs[0]):
        ep = episodes[idx].copy()
        ep["similarity_score"] = float(1 / (1 + dist))
        results.append(ep)
    return results


def retrieve_similar_episodes_embargoed(current_episode, k=5, min_days_away=30):
    query_vec = build_embedding(current_episode)
    query_vec = scaler.transform([query_vec]).astype("float32")
    distances, idxs = index.search(query_vec, k * 50)
    results = []
    for dist, idx in zip(distances[0], idxs[0]):
        ep = episodes[idx]
        gap_days = (current_episode["date"] - ep["date"]).days  # SIGNED, not abs(): an episode dated
        # AFTER the query must always fail this check, regardless of how far away it is
        if gap_days < min_days_away:
            continue
        e = ep.copy()
        e["similarity_score"] = float(1 / (1 + dist))
        results.append(e)
        if len(results) == k:
            break
    return results


def format_episodes_for_llama(retrieved_episodes):
    lines = ["### Historical Similar Episodes:\n"]
    for i, ep in enumerate(retrieved_episodes, 1):
        date = ep["date"].strftime("%Y-%m-%d") if hasattr(ep["date"], "strftime") else str(ep["date"])
        score = ep["similarity_score"]
        policy = "YES" if ep["policy_event"] else "NO"
        outcomes = ep["outcome_returns"]
        lines.append(
            f"Episode {i} | Date: {date} | Similarity: {score:.2f}\n"
            f"  Policy Shock: {policy}\n"
            f"  {FORECAST_HORIZON}-day Outcomes: EURUSD={outcomes[0]:+.3f}, GBPUSD={outcomes[1]:+.3f}, "
            f"GOLD={outcomes[2]:+.3f}, OIL={outcomes[3]:+.3f}\n"
        )
    return "\n".join(lines)
````

## Section 4 — Rebalance dates + build a query episode for each

Runs the pipeline at **every monthly rebalance date** in your test period, not just the most
recent day — this is what Step 6/7's backtest actually needs. `TEST_START`/`TEST_END` below are
placeholders — set them to match your team's real train/test split (Step 1) and Step 7's backtest
schedule exactly, or you'll be generating views for dates nobody else is using.


````python
df = pd.read_csv("project8_dataset_v2.csv", parse_dates=["Date"], index_col="Date")
df[assets] = df[assets].ffill()

policy_df = pd.read_csv("policy_events.csv", parse_dates=["date"])

# --- Rebalance dates: read them from Step 3's own output instead of recomputing them here.
#     Recomputing independently (via TEST_START/TEST_END + freq="ME") is exactly how a 1-date
#     mismatch (18 vs. 19 dates) crept in previously. Reading them directly from
#     lag_llama_views.csv guarantees Step 3 and Step 5 always agree, even if Step 3's own
#     date logic changes later. ---
if LAGLLAMA_IS_PER_DATE:
    candidate_dates = sorted(lagllama_views["Date"].unique())
else:
    # fallback only used if Step 3's per-date file truly isn't available yet
    TEST_START, TEST_END = "2025-01-01", "2026-06-30"
    candidate_dates = pd.date_range(TEST_START, TEST_END, freq="ME")

REBALANCE_DATES = []
for target in candidate_dates:
    target = pd.Timestamp(target)
    available = df.index[df.index <= target]
    if len(available) >= WINDOW_SIZE + 2:       # need enough prior history for a full window
        REBALANCE_DATES.append(available[-1])   # snap to the last real trading day on/before target

if REBALANCE_DATES:
    print(f"{len(REBALANCE_DATES)} rebalance dates: {REBALANCE_DATES[0].date()} -> {REBALANCE_DATES[-1].date()}")
else:
    print("0 rebalance dates found -- check lag_llama_views.csv / your data range")


def build_query_episode(query_date):
    """Same shape as Step 4B's stored episodes, but anchored at an arbitrary rebalance date
    instead of always 'today' -- this is what makes the loop in Section 8 possible."""
    i = df.index.get_loc(query_date)
    window = df.iloc[i - (WINDOW_SIZE + 1): i]
    price_returns = window[assets].pct_change().dropna().values.flatten()
    macro_features = window[macro_cols].iloc[1:].mean().values
    recent_policy_row = policy_df[policy_df["date"] <= query_date].sort_values("date").tail(1)
    policy_event_flag = int(len(recent_policy_row) > 0 and
                             (query_date - recent_policy_row["date"].iloc[0]).days <= FORECAST_HORIZON)
    query_episode = {
        "date": query_date,
        "price_returns": price_returns,
        "macro_features": macro_features,
        "policy_event": policy_event_flag,
        "outcome_returns": np.zeros(len(assets)),
    }
    return query_episode, recent_policy_row
````

## Section 5 — Retrieve historical analogs (this IS Step 4B, called from 4C)

Wrapped as a function now so it can run once per rebalance date. Using the **embargoed**
retrieval by default — per your teammate's note, plain retrieval tends to just return
calendar-adjacent days, which isn't a useful "have we seen this before" analog.


````python
def get_memory_context(query_episode):
    retrieved = retrieve_similar_episodes_embargoed(query_episode, k=5, min_days_away=30)
    memory_context = format_episodes_for_llama(retrieved)
    avg_similarity = np.mean([r["similarity_score"] for r in retrieved]) if retrieved else 0.0
    return retrieved, memory_context, avg_similarity
````

## Section 6 — Prompt construction

Two design choices worth explaining:

- **Structured JSON output, strictly typed.** LLMs asked for free-text market commentary
  are hard to parse reliably into numbers. Asking for JSON with a fixed schema (and
  validating it in Section 7) avoids fragile regex-parsing of prose.
- **Self-consistency for confidence** (per Lee et al. 2025, "LLM-Enhanced Black-Litterman
  Portfolio Optimization"): rather than trusting one LLM response, call it several times
  at nonzero temperature and use the *spread* of its own answers as the confidence
  signal — mirroring exactly how Lag-Llama got its confidence in Step 3 (variance across
  100 samples). This also gives you a defensible, citable methodology for your report.


````python
SYSTEM_PROMPT = f"""You are a quantitative macro analyst producing structured inputs for a
Black-Litterman portfolio optimizer. You will be given:
  (1) A pretrained time-series model's forecast for each asset over the next {FORECAST_HORIZON} trading days,
      with its own confidence score.
  (2) The most recent policy/news event, if any.
  (3) Historical episodes retrieved from memory that are structurally similar to the current
      market window, each showing what actually happened to each asset {FORECAST_HORIZON} days later.

Your job: produce a FINAL {FORECAST_HORIZON}-day expected return and confidence (0 to 1) for each asset,
reasoning about whether the historical analogs support, contradict, or are irrelevant to the
time-series model's forecast. If the retrieved historical analogs have low similarity scores
(below ~0.05), say so explicitly and rely more on general reasoning about the policy event
rather than forcing a pattern that doesn't exist.

IMPORTANT: expected_return must be a decimal fraction, not a percentage.
Example: a +2.5% return must be written as 0.025, not 2.5.
Values should realistically fall between -0.30 and +0.30 for a {FORECAST_HORIZON}-day horizon
(widened from the old +/-0.20 5-day bound -- a ~monthly-scale move, e.g. Oil's ~17% historical
monthly std, needs more room than a weekly one did).

Respond with ONLY valid JSON in this exact schema, nothing else, no markdown fences:
{{
  "EURUSD": {{"expected_return": <float>, "confidence": <float 0-1>, "reasoning": "<one sentence>"}},
  "GBPUSD": {{"expected_return": <float>, "confidence": <float 0-1>, "reasoning": "<one sentence>"}},
  "GOLD":   {{"expected_return": <float>, "confidence": <float 0-1>, "reasoning": "<one sentence>"}},
  "OIL":    {{"expected_return": <float>, "confidence": <float 0-1>, "reasoning": "<one sentence>"}}
}}"""

def get_lagllama_forecast_for_date(query_date):
    """Looks up Step 3's forecast for this date if lag_llama_views.csv is per-date;
    otherwise reuses the single static forecast for every date (see the WARNING printed
    in Section 2 — confirm with your Step 3 teammate whether that's good enough)."""
    if LAGLLAMA_IS_PER_DATE:
        rows = lagllama_views[lagllama_views["Date"] == query_date]
        if len(rows) == 0:
            earlier = lagllama_views[lagllama_views["Date"] <= query_date]
            if len(earlier) > 0:
                rows = earlier[earlier["Date"] == earlier["Date"].max()]
            else:
                rows = lagllama_views
    else:
        rows = lagllama_views
    return rows


def build_user_prompt(query_date, lagllama_rows, recent_policy_row, memory_context, avg_similarity):
    lines = [f"### Time-Series Model Forecast (Step 3, Lag-Llama) as of {query_date.date()}:\n"]
    for _, row in lagllama_rows.iterrows():
        lines.append(
            f"- {row['Asset_clean']}: expected {FORECAST_HORIZON}-day return {row['Expected_Return']:+.4f}, "
            f"model confidence {row['Confidence']:.2f} ({row['Signal']})"
        )

    lines.append("\n### Most Recent Policy Event:")
    if len(recent_policy_row) > 0:
        r = recent_policy_row.iloc[0]
        lines.append(f"- {r['date'].date()} | {r['type']} | severity={r['severity']} | {r['description']}")
    else:
        lines.append("- No recent policy event on record near this date.")

    lines.append(f"\n{memory_context}")
    lines.append(f"\n(Average similarity of retrieved analogs: {avg_similarity:.3f} — "
                  f"{'treat as a weak/uninformative analog set' if avg_similarity < 0.05 else 'reasonably informative'})")
    return "\n".join(lines)
````

## Section 7 — Call Llama-3 with self-consistency, parse robustly

Same validation logic as before, now wrapped in `run_self_consistency()` so Section 8 can call
it once per rebalance date. A short `time.sleep()` between attempts helps avoid Groq rate limits
once you're making dozens of calls across many dates — safe to shorten/remove on a paid tier.


````python
N_SAMPLES = 3  # reduced from 5 -- cuts total API calls ~40%, still enough for self-consistency
ASSET_KEYS = ["EURUSD", "GBPUSD", "GOLD", "OIL"]

def call_llama(user_prompt, temperature=0.7, max_retries=3):
    """Retries with exponential backoff specifically on rate-limit errors, instead of just
    recording a failed attempt and moving on -- this is what was causing dates to get
    SKIPPED partway through a full run once enough calls had been made to hit Groq's limit."""
    for retry in range(max_retries):
        try:
            response = client.chat.completions.create(
                model=MODEL_NAME,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": user_prompt},
                ],
                temperature=temperature,
                max_tokens=2000,
                reasoning_effort="low",
            )
            msg = response.choices[0].message
            text = (msg.content or "").strip()
            if not text:
                reasoning_len = len(getattr(msg, "reasoning", "") or "")
                raise ValueError(
                    f"Empty content (finish_reason={response.choices[0].finish_reason}, "
                    f"reasoning_chars={reasoning_len}) -- model likely ran out of max_tokens while reasoning."
                )
            text = text.replace("```json", "").replace("```", "").strip()
            return json.loads(text)
        except Exception as e:
            is_rate_limit = "rate" in str(e).lower() or "429" in str(e)
            if is_rate_limit and retry < max_retries - 1:
                wait = 2 ** (retry + 2)   # 4s, 8s, 16s
                print(f"      rate limited, waiting {wait}s (retry {retry + 1}/{max_retries})...")
                time.sleep(wait)
                continue
            raise


def is_valid(parsed):
    if not isinstance(parsed, dict):
        return False
    for k in ASSET_KEYS:
        if k not in parsed:
            return False
        v = parsed[k]
        if "expected_return" not in v or "confidence" not in v:
            return False
        if not (0 <= float(v["confidence"]) <= 1):
            return False
        if abs(float(v["expected_return"])) > 1.0:
            return False
    return True


def run_self_consistency(user_prompt, n_samples=N_SAMPLES):
    samples = []
    for attempt in range(n_samples):
        try:
            parsed = call_llama(user_prompt, temperature=0.7)
            if is_valid(parsed):
                samples.append(parsed)
            else:
                print(f"    attempt {attempt}: failed validation, discarded")
        except Exception as e:
            print(f"    attempt {attempt}: {type(e).__name__}: {e}")
        time.sleep(1.0)   # slightly longer gap between calls to reduce how often the limit is hit at all
    return samples
````

## Section 8 — Run every rebalance date, aggregate into final views

This is the actual loop your team needs: for each date in `REBALANCE_DATES`, build the query
episode, retrieve analogs, build the prompt, run self-consistency, and aggregate — same
aggregation math as before (mean return, spread-based confidence penalty), just repeated per
date. Output columns match what your team expects: `Date, Asset, Signal, Expected_Return_Pct,
Expected_Return, Confidence, Reasoning`.


````python
import os

NEUTRAL_BAND = 0.02  # +/- 2% dead zone for Signal classification (widened from 0.5% used at the
# 5-day horizon -- a 30-day return needs a wider dead zone before "noise" becomes "signal";
# tune if your team uses a different cutoff)

def signal_from_return(r):
    if r > NEUTRAL_BAND:
        return "BULLISH"
    elif r < -NEUTRAL_BAND:
        return "BEARISH"
    return "NEUTRAL"


CHECKPOINT_FILE = "final_views_checkpoint.csv"

# --- Resume support: if a checkpoint from a previous (possibly rate-limited) run exists,
#     load it and skip any dates already completed. Re-running this cell -- immediately, or
#     after waiting out a rate limit, even the next day -- picks up exactly where it left off.
#     No manual date-range splitting or merging required. ---
if os.path.exists(CHECKPOINT_FILE):
    checkpoint_df = pd.read_csv(CHECKPOINT_FILE, parse_dates=["Date"])
    done_dates = set(checkpoint_df["Date"].dt.date)
    print(f"Resuming: {len(done_dates)} dates already completed in a previous run, will be skipped.")
else:
    checkpoint_df = pd.DataFrame()
    done_dates = set()

# rebuild per_date_views for already-completed dates too, so Section 10's JSON output is
# complete even on a resumed run without needing to regenerate already-done dates
per_date_views = {}
if len(checkpoint_df) > 0:
    for date_val, g in checkpoint_df.groupby(checkpoint_df["Date"].dt.date):
        per_date_views[str(date_val)] = {
            "views": dict(zip(g["Asset"], g["Expected_Return"])),
            "confidences": g["Confidence"].tolist(),
        }

remaining_dates = [d for d in REBALANCE_DATES if d.date() not in done_dates]
print(f"{len(remaining_dates)} of {len(REBALANCE_DATES)} dates remaining this run")

for d_i, query_date in enumerate(remaining_dates):
    print(f"[{d_i + 1}/{len(remaining_dates)}] {query_date.date()}")

    query_episode, recent_policy_row = build_query_episode(query_date)
    retrieved, memory_context, avg_similarity = get_memory_context(query_episode)
    lagllama_rows = get_lagllama_forecast_for_date(query_date)
    user_prompt = build_user_prompt(query_date, lagllama_rows, recent_policy_row, memory_context, avg_similarity)

    samples = run_self_consistency(user_prompt)
    print(f"    {len(samples)}/{N_SAMPLES} valid responses")
    if len(samples) < 2:
        print(f"    SKIPPING {query_date.date()} -- too few valid LLM responses (will retry on next run)")
        continue

    date_rows = []
    date_views = {}
    date_confidences = []
    for asset in ASSET_KEYS:
        returns_i = np.array([s[asset]["expected_return"] for s in samples])
        stated_conf_i = np.array([s[asset]["confidence"] for s in samples])

        mean_return = float(returns_i.mean())
        spread = float(returns_i.std())
        mean_stated_conf = float(stated_conf_i.mean())
        consistency_penalty = 1 / (1 + 50 * spread)
        final_conf = float(np.clip(mean_stated_conf * consistency_penalty, 0.01, 0.99))

        date_views[asset] = mean_return
        date_confidences.append(final_conf)

        date_rows.append({
            "Date": query_date.date(),
            "Asset": asset,
            "Signal": signal_from_return(mean_return),
            "Expected_Return_Pct": round(mean_return * 100, 2),
            "Expected_Return": mean_return,
            "Confidence": final_conf,
            "Reasoning": samples[0][asset]["reasoning"],
            "Self_Consistency_Spread": spread,
            "Mean_Stated_Confidence": mean_stated_conf,
        })

    per_date_views[str(query_date.date())] = {"views": date_views, "confidences": date_confidences}

    # checkpoint immediately -- this date's work is now safe even if the very next date hits
    # a rate limit the run can't recover from
    date_df = pd.DataFrame(date_rows)
    write_header = not os.path.exists(CHECKPOINT_FILE)
    date_df.to_csv(CHECKPOINT_FILE, mode="a", header=write_header, index=False)

# reload the full checkpoint (previously-done + just-completed dates) as the working result set
result_df = pd.read_csv(CHECKPOINT_FILE, parse_dates=["Date"])
n_done = result_df["Date"].nunique()
print(f"\nProgress: {n_done} of {len(REBALANCE_DATES)} rebalance dates completed so far")
if result_df["Date"].nunique() < len(REBALANCE_DATES):
    print("Not all dates finished this run -- just re-run this cell again (immediately, or after")
    print("waiting out a rate limit) and it will pick up only the remaining dates.")
result_df
````

## Section 9 — Spot-check against Step 3 (most recent rebalance date)

A full multi-date comparison only makes sense once Step 3 also produces per-date forecasts (see
the WARNING in Section 2). Until then, this spot-checks just the most recent rebalance date —
where the LLM should plausibly move things is GOLD/OIL, where Lag-Llama's own confidence was
near-zero because 2025 prices were outside its training range.


````python
latest_date = result_df["Date"].max()
latest = result_df[result_df["Date"] == latest_date].set_index("Asset")

compare = lagllama_views.set_index("Asset_clean")[["Expected_Return", "Confidence"]].copy()
compare.columns = ["LagLlama_Return", "LagLlama_Confidence"]
compare["Llama3_Return"] = compare.index.map(latest["Expected_Return"])
compare["Llama3_Confidence"] = compare.index.map(latest["Confidence"])
print(f"Spot-check for {latest_date}:")
print(compare.round(4))
````
