# Step 6, Part B — Online RL: PPO-Lagrangian
### Step 5 of the guide: the Lagrangian safety loop — revised for stable constrained training

This notebook continues from `Step6B_PlainPPO_SanityCheck.ipynb` and loads `ppo_plain_finetuned.zip`.

The previous run demonstrated that the safety mechanism worked, but it paid too much return: **+51.85% → +36.36%**. The reason is visible in the implementation: the training cost is defined as a non-negative hinge (`max(0, excess)`), and the dual update adds that non-negative quantity every round. Consequently λ can only rise; once the policy becomes safer, the multiplier cannot relax, so later rounds keep increasing safety pressure and the return keeps deteriorating.

The revised loop keeps the original hard risk budgets intact:

- CVaR(5%) budget = **2.3%**;
- maximum drawdown budget = **19%**.

But the dual update is changed to a proper constrained feedback signal that can move **up or down**, using small training tolerances on violation frequency: 5% of days for CVaR and 1% for drawdown. The penalty itself still uses the *magnitude* of the CVaR/drawdown shortfall, so larger breaches remain more expensive than tiny ones.

Training is also made gentler: one PPO rollout per round, lower learning rate, fewer PPO epochs, tighter clipping, and a KL stop. A feasible checkpoint with the highest TRAIN-period reward is retained instead of automatically using the last round.


## Setup — rebuild the environment exactly as before, load the plain-PPO-finetuned policy

Identical data loading and `PortfolioAdjustEnv` definition to `Step6A_WarmStart_PPO.ipynb` and
`Step6B_PlainPPO_SanityCheck.ipynb` (TRAIN, 2018-01-03 to 2024-12-31, 1,758 days, same 12-dim
state, `cvar_budget=0.023` / `drawdown_budget=0.19`, recalibrated in Part A from the anchor
policy's own real risk distribution over TRAIN). Rebuilding it here rather than importing it keeps
this notebook self-contained and runnable on its own, matching the pattern of every notebook so
far.


```python
!pip install --quiet --break-system-packages gymnasium stable-baselines3
# torch/pandas/numpy ship preinstalled in Colab; uncomment below if running in a bare local
# environment instead:
# !pip install --quiet --break-system-packages torch pandas numpy matplotlib
```

```python
import pickle
import numpy as np
import pandas as pd
import gymnasium as gym
from stable_baselines3 import PPO

np.random.seed(42)

with open("offline_trajectories.pkl", "rb") as f:
    saved = pickle.load(f)
ASSETS = saved["assets"]
STATE_DIM = saved["state_dim"]

df = pd.read_csv("project8_dataset_v2.csv", parse_dates=["Date"], index_col="Date")
prices = df[[f"{a}_Close" for a in ASSETS]].copy()
prices.columns = ASSETS
prices = prices.dropna()
returns = prices.pct_change().dropna()
returns_clean = returns.clip(lower=-0.20, upper=0.20)

train_dates = df[df["PERIOD"] == "TRAIN"].index
train_rets = returns_clean.loc[returns_clean.index.isin(train_dates)]

anchor = pd.read_csv("bl_baseline_train_weights.csv", parse_dates=["Date"]).set_index("Date")
daily_anchor = anchor.reindex(train_rets.index, method="ffill").bfill()
assert len(train_rets) == 1758

CVAR_WINDOW = 60
N_STEPS = len(train_rets) - CVAR_WINDOW - 1


def build_state_features(t_idx):
    if t_idx < 5:
        trailing = train_rets.iloc[:t_idx] if t_idx > 0 else train_rets.iloc[:1] * 0
    else:
        trailing = train_rets.iloc[t_idx - 5:t_idx]
    mean_ret = trailing[ASSETS].mean().values if len(trailing) > 0 else np.zeros(len(ASSETS))
    vol = trailing[ASSETS].std().fillna(0).values if len(trailing) > 0 else np.zeros(len(ASSETS))
    w = daily_anchor.iloc[t_idx][ASSETS].values
    return np.concatenate([mean_ret, vol, w]).astype(np.float32)


def compute_cvar(returns_arr, alpha=0.05):
    """Historical-simulation CVaR: average loss in the worst alpha fraction of periods."""
    losses = -returns_arr
    var_threshold = np.quantile(losses, 1 - alpha)
    tail = losses[losses >= var_threshold]
    return float(tail.mean()) if len(tail) > 0 else float(losses.max())


class PortfolioAdjustEnv(gym.Env):
    # identical to Step6A_WarmStart_PPO.ipynb / Step6B_PlainPPO_SanityCheck.ipynb, including the
    # TRAIN-recalibrated CVaR + drawdown budgets
    def __init__(self, daily_anchor, train_rets, assets, state_fn, cvar_budget=0.023,
                 cvar_window=60, cvar_alpha=0.05, drawdown_budget=0.19):
        super().__init__()
        self.daily_anchor = daily_anchor
        self.train_rets = train_rets
        self.assets = assets
        self.state_fn = state_fn
        self.cvar_budget = cvar_budget
        self.cvar_window = cvar_window
        self.cvar_alpha = cvar_alpha
        self.drawdown_budget = drawdown_budget
        self.n_assets = len(assets)
        self.state_dim = state_fn(cvar_window).shape[0]

        self.action_space = gym.spaces.Box(low=-0.15, high=0.15, shape=(self.n_assets,), dtype=np.float32)
        self.observation_space = gym.spaces.Box(low=-np.inf, high=np.inf, shape=(self.state_dim,), dtype=np.float32)
        self.t = None
        self.cum_value = None
        self.peak_value = None

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        self.t = self.cvar_window
        self.cum_value = 1.0
        self.peak_value = 1.0
        return self.state_fn(self.t), {}

    def step(self, action):
        anchor_w = self.daily_anchor.iloc[self.t][self.assets].values
        adjusted = np.clip(anchor_w + np.asarray(action), 0, 1)
        adjusted = adjusted / adjusted.sum()

        reward = float(adjusted @ self.train_rets.iloc[self.t][self.assets].values)

        window_rets = self.train_rets.iloc[self.t - self.cvar_window:self.t][self.assets].values @ adjusted
        cvar = compute_cvar(window_rets, self.cvar_alpha)
        cvar_cost = max(0.0, cvar - self.cvar_budget)

        self.cum_value *= (1 + reward)
        self.peak_value = max(self.peak_value, self.cum_value)
        drawdown = (self.cum_value / self.peak_value) - 1
        drawdown_cost = max(0.0, abs(drawdown) - self.drawdown_budget)

        total_cost = cvar_cost + drawdown_cost

        self.t += 1
        terminated = self.t >= len(self.train_rets) - 1
        obs = self.state_fn(self.t) if not terminated else np.zeros(self.state_dim, dtype=np.float32)
        info = {"cost": total_cost, "cvar": cvar, "cvar_cost": cvar_cost,
                "drawdown": drawdown, "drawdown_cost": drawdown_cost,
                "portfolio_value": self.cum_value}
        return obs, reward, terminated, False, info


base_env = PortfolioAdjustEnv(daily_anchor, train_rets, ASSETS, build_state_features)
model = PPO.load("ppo_plain_finetuned", env=base_env)
print("Environment rebuilt on TRAIN, plain-PPO-finetuned policy loaded from ppo_plain_finetuned.zip")
print(f"CVaR budget: {base_env.cvar_budget}  |  Drawdown budget: {base_env.drawdown_budget}")
print(f"TRAIN-period episode length: {N_STEPS} days")
```

## Evaluation helper

The helper now records the separate CVaR and drawdown costs, violation rates, maximum drawdown, and mean action magnitude. This is important because the two constraints should not be collapsed into one irreversible λ signal.


```python
def run_episode(model, env):
    """Deterministic TRAIN rollout used for dual updates and checkpoint selection."""
    obs, _ = env.reset()
    total_reward = 0.0
    cvar_costs, dd_costs, actions, drawdowns, cvars = [], [], [], [], []
    cvar_violations, dd_violations = 0, 0
    terminated = False

    while not terminated:
        action, _ = model.predict(obs, deterministic=True)
        actions.append(np.asarray(action).copy())
        obs, reward, terminated, truncated, info = env.step(action)
        total_reward += reward
        cvar_costs.append(info["cvar_cost"])
        dd_costs.append(info["drawdown_cost"])
        cvars.append(info["cvar"])
        drawdowns.append(info["drawdown"])
        cvar_violations += int(info["cvar_cost"] > 0)
        dd_violations += int(info["drawdown_cost"] > 0)

    n = max(1, len(cvar_costs))
    return {
        "reward": float(total_reward),
        "avg_cost": float(np.mean(np.asarray(cvar_costs) + np.asarray(dd_costs))),
        "avg_cvar_cost": float(np.mean(cvar_costs)),
        "avg_dd_cost": float(np.mean(dd_costs)),
        "cvar_violations": int(cvar_violations),
        "dd_violations": int(dd_violations),
        "cvar_violation_rate": float(cvar_violations / n),
        "dd_violation_rate": float(dd_violations / n),
        "max_cvar": float(np.max(cvars)),
        "max_drawdown": float(-np.min(drawdowns)),
        "mean_abs_action": float(np.mean(np.abs(np.asarray(actions)))),
    }
```

## Baseline — evaluate the plain-PPO-finetuned policy BEFORE any Lagrangian training

This is the "before" number for everything that follows: the policy Step 4 already confirmed
learns well, but which has never once been trained against its own constraint violations.


```python
baseline = run_episode(model, base_env)
print("BEFORE Lagrangian training (plain-PPO-finetuned policy):")
print(f"  Total episode reward:                  {baseline['reward']:+.2%}")
print(f"  Average CVaR shortfall:                {baseline['avg_cvar_cost']:.6f}")
print(f"  Average drawdown shortfall:            {baseline['avg_dd_cost']:.6f}")
print(f"  CVaR budget violations:                {baseline['cvar_violations']} / {N_STEPS} "
      f"({baseline['cvar_violation_rate']:.2%})")
print(f"  Drawdown budget violations:            {baseline['dd_violations']} / {N_STEPS} "
      f"({baseline['dd_violation_rate']:.2%})")
print(f"  Maximum drawdown:                      {baseline['max_drawdown']:.2%}")
print(f"  Mean |action| magnitude:               {baseline['mean_abs_action']:.4f}")
```

## Diagnose the reward/cost scale — before touching `lam_lr`

The guide's own "Common pitfalls" section flags this exact failure mode: *"if daily returns are
~0.001 and cost is ~0.05, the penalty term dominates immediately and the agent just freezes...
consider scaling reward or cost... before tuning `lam_lr`."* Rather than assume `lam_lr=20.0`
still applies unchanged just because it worked on the old TEST-period setup, this re-measures the
actual per-step reward and cost distribution under TRAIN — the same "calibrate from the real
observed distribution, not round numbers" approach Part A already used for `cvar_budget` and
`drawdown_budget`, applied here to the one parameter that genuinely can't be recalibrated without
a real trained policy to measure (unlike the budgets, which the zero-action rollout in Part A
could determine on its own).


```python
# Diagnose the two constraints separately
obs, _ = base_env.reset()
terminated = False
rows = []
while not terminated:
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = base_env.step(action)
    rows.append({"reward": reward, "cvar": info["cvar"], "cvar_cost": info["cvar_cost"],
                 "drawdown": info["drawdown"], "drawdown_cost": info["drawdown_cost"]})

diag = pd.DataFrame(rows)
print(f"Mean daily reward:       {diag['reward'].mean():+.6f}")
print(f"Mean CVaR:               {diag['cvar'].mean():.6f}")
print(f"Mean CVaR shortfall:     {diag['cvar_cost'].mean():.6f}")
print(f"Mean DD shortfall:       {diag['drawdown_cost'].mean():.6f}")
print(f"Max drawdown:            {-diag['drawdown'].min():.2%}")
print(f"CVaR violation rate:     {(diag['cvar_cost'] > 0).mean():.2%}")
print(f"DD violation rate:       {(diag['drawdown_cost'] > 0).mean():.2%}")
print("\nThe important correction is that the dual update below uses violation-rate minus target, so lambda can decrease once the policy is sufficiently safe.")
```

## Lagrangian design used below

The original loop used a single monotonic multiplier:

`lambda <- max(0, lambda + lr * nonnegative_cost)`.

That is the source of the late-round over-penalization. The revised loop uses two multipliers:

- `lambda_cvar` for CVaR shortfall;
- `lambda_dd` for drawdown shortfall.

The dual constraints are practical violation-frequency targets:

- CVaR violation rate target = 5%;
- drawdown violation rate target = 1%.

Thus each multiplier can move up **or down**. The underlying 2.3% CVaR and 19% drawdown budgets are unchanged.


## The Lagrangian wrapper

Exactly the guide's `LambdaPenalizedEnv`: wraps the environment so training sees
`reward - lambda * cost` instead of raw `reward`. Keeping this separate from `PortfolioAdjustEnv`
itself (rather than baking the penalty into the environment) is what lets `lambda` be swapped in
between rounds without rebuilding the environment, and keeps `info["cost"]` available unpenalized
for evaluation and monitoring throughout.


```python
class LambdaPenalizedEnv(gym.Wrapper):
    """Reward = raw return - separate adaptive penalties for CVaR/DD shortfalls."""
    def __init__(self, env, lam_cvar=0.0, lam_dd=0.0):
        super().__init__(env)
        self.lam_cvar = lam_cvar
        self.lam_dd = lam_dd

    def step(self, action):
        obs, reward, term, trunc, info = self.env.step(action)
        penalty = self.lam_cvar * info["cvar_cost"] + self.lam_dd * info["drawdown_cost"]
        penalized_reward = reward - penalty
        return obs, penalized_reward, term, trunc, info
```

## The Lagrangian safety loop

Dual ascent on `lambda`, exactly the guide's update rule: after each round of training,
`lambda <- max(0, lambda + lam_lr * (avg_cost - cost_budget))`. `cost_budget=0.0` targets
(near-)zero constraint violation, matching the guide's default — the two components of cost
(CVaR shortfall and drawdown shortfall) are already recalibrated in Part A to bind only
occasionally under a *reasonable* anchor policy over TRAIN, so driving the combined cost toward
zero is a genuinely meaningful target here, not an impossible one.


```python
# Conservative constrained PPO loop
from stable_baselines3.common.utils import get_schedule_fn

LEARNING_RATE = 3e-5
N_EPOCHS = 3
CLIP_RANGE = 0.10
TARGET_KL = 0.01
STEPS_PER_ROUND = 2048      # exactly one SB3 rollout per dual update
N_ROUNDS = 12

# Start with a small safety pressure immediately; do not give round 0 an unpenalized drift period.
lam_cvar = 0.02
lam_dd = 0.02
LAM_LR_CVAR = 0.10
LAM_LR_DD = 0.20
CVAR_TARGET_RATE = 0.05
DD_TARGET_RATE = 0.01

wrapped_env = LambdaPenalizedEnv(base_env, lam_cvar=lam_cvar, lam_dd=lam_dd)
model.set_env(wrapped_env)
model.learning_rate = LEARNING_RATE
model.lr_schedule = get_schedule_fn(LEARNING_RATE)
model.clip_range = get_schedule_fn(CLIP_RANGE)
model.n_epochs = N_EPOCHS
model.target_kl = TARGET_KL
for group in model.policy.optimizer.param_groups:
    group["lr"] = LEARNING_RATE

history = [{
    "round": -1,
    "lambda_cvar": lam_cvar,
    "lambda_dd": lam_dd,
    "reward": baseline["reward"],
    "avg_cost": baseline["avg_cost"],
    "avg_cvar_cost": baseline["avg_cvar_cost"],
    "avg_dd_cost": baseline["avg_dd_cost"],
    "cvar_violations": baseline["cvar_violations"],
    "dd_violations": baseline["dd_violations"],
    "cvar_violation_rate": baseline["cvar_violation_rate"],
    "dd_violation_rate": baseline["dd_violation_rate"],
    "max_drawdown": baseline["max_drawdown"],
    "mean_abs_action": baseline["mean_abs_action"],
}]

best_feasible_path = None
best_feasible_reward = -np.inf
best_any = None
best_any_key = None

for round_i in range(N_ROUNDS):
    wrapped_env.lam_cvar = lam_cvar
    wrapped_env.lam_dd = lam_dd
    model.learn(total_timesteps=STEPS_PER_ROUND, reset_num_timesteps=False)

    res = run_episode(model, base_env)

    # Proper dual feedback: the multiplier can rise or fall around the target violation rate.
    lam_cvar = max(0.0, lam_cvar + LAM_LR_CVAR * (res["cvar_violation_rate"] - CVAR_TARGET_RATE))
    lam_dd = max(0.0, lam_dd + LAM_LR_DD * (res["dd_violation_rate"] - DD_TARGET_RATE))

    row = {
        "round": round_i,
        "lambda_cvar": lam_cvar,
        "lambda_dd": lam_dd,
        **res,
    }
    history.append(row)

    feasible = (
        res["cvar_violation_rate"] <= CVAR_TARGET_RATE and
        res["dd_violation_rate"] <= DD_TARGET_RATE and
        res["avg_cost"] < baseline["avg_cost"]
    )

    # Keep the highest-return feasible policy, not automatically the last policy.
    if feasible and res["reward"] > best_feasible_reward:
        best_feasible_reward = res["reward"]
        best_feasible_path = f"ppo_lagrangian_candidate_r{round_i}"
        model.save(best_feasible_path)

    # Fallback ranking if no checkpoint reaches both practical safety targets.
    safety_key = (
        res["cvar_violation_rate"] / max(CVAR_TARGET_RATE, 1e-9) +
        res["dd_violation_rate"] / max(DD_TARGET_RATE, 1e-9) +
        res["avg_cost"] / max(baseline["avg_cost"], 1e-9)
    )
    if best_any is None or safety_key < best_any_key or (abs(safety_key - best_any_key) < 1e-12 and res["reward"] > best_any["reward"]):
        best_any = res.copy()
        best_any_key = safety_key
        model.save("ppo_lagrangian_safest_candidate")

    print(f"round {round_i:2d}: reward={res['reward']:+.2%}  avg_cost={res['avg_cost']:.6f}  "
          f"lambda_cvar={lam_cvar:.4f}  lambda_dd={lam_dd:.4f}  "
          f"cvar_v={res['cvar_violations']}/{N_STEPS} ({res['cvar_violation_rate']:.2%})  "
          f"dd_v={res['dd_violations']}/{N_STEPS} ({res['dd_violation_rate']:.2%})  "
          f"maxDD={res['max_drawdown']:.2%}  {'<-- feasible' if feasible else ''}")

history_df = pd.DataFrame(history)
```

## What to watch during the loop

The important signals are now:

- CVaR and drawdown violation rates should move toward the 5% / 1% training targets.
- `lambda_cvar` and `lambda_dd` should remain bounded and be able to move both up and down.
- Return should not show the monotonic late-round collapse seen in the previous run.
- The selected checkpoint is chosen by **highest return among sufficiently safe policies**, rather than using the last round.
- The final Step 7 walk-forward evaluation remains the out-of-sample performance test.


```python
# Verify the selected policy still makes non-trivial adjustments
if best_feasible_path is not None:
    selected_path = best_feasible_path
    selection_reason = "highest-return checkpoint satisfying both practical safety-rate targets"
else:
    selected_path = "ppo_lagrangian_safest_candidate"
    selection_reason = "safest available checkpoint because no round satisfied both targets"

selected_model = PPO.load(selected_path, env=base_env)
selected = run_episode(selected_model, base_env)

print(f"Selected checkpoint: {selected_path}")
print(f"Selection rule:      {selection_reason}")
print(f"Reward:               {selected['reward']:+.2%}  (baseline {baseline['reward']:+.2%})")
print(f"Avg total cost:       {selected['avg_cost']:.6f} (baseline {baseline['avg_cost']:.6f})")
print(f"CVaR violations:     {selected['cvar_violations']}/{N_STEPS} ({selected['cvar_violation_rate']:.2%})")
print(f"DD violations:       {selected['dd_violations']}/{N_STEPS} ({selected['dd_violation_rate']:.2%})")
print(f"Maximum drawdown:     {selected['max_drawdown']:.2%}")
print(f"Mean |action|:        {selected['mean_abs_action']:.4f} (baseline {baseline['mean_abs_action']:.4f})")

assert selected["avg_cost"] < baseline["avg_cost"], "Selected policy did not reduce the safety cost."
assert selected["dd_violations"] <= baseline["dd_violations"], "Selected policy worsened drawdown violations." 
assert selected["mean_abs_action"] > 0.5 * baseline["mean_abs_action"], "Possible do-nothing collapse." 
print("\nPASS: selected policy reduces safety cost, does not worsen drawdown violations, and remains an active adjustment policy.")
```

## Validate: did the Lagrangian mechanism actually make the policy safer?

The thing this step specifically needs to demonstrate — not "did reward improve" (that was
Step 4's job), but "did the trained-against cost signal actually go down." Compared against the
baseline captured before this notebook's training loop started.


```python
final = selected

print(f"Average per-step cost: {baseline['avg_cost']:.6f} (baseline) -> {final['avg_cost']:.6f} (selected)")
print(f"CVaR violations:       {baseline['cvar_violations']}/{N_STEPS} -> {final['cvar_violations']}/{N_STEPS}")
print(f"Drawdown violations:   {baseline['dd_violations']}/{N_STEPS} -> {final['dd_violations']}/{N_STEPS}")
print(f"Episode reward:         {baseline['reward']:+.2%} -> {final['reward']:+.2%}")
print(f"Maximum drawdown:       {baseline['max_drawdown']:.2%} -> {final['max_drawdown']:.2%}")

assert final["avg_cost"] < baseline["avg_cost"], "Safety cost did not fall."
assert final["dd_violations"] <= baseline["dd_violations"], "Drawdown violations worsened."
print("\nPASS: CVaR and drawdown are both explicitly included in the constrained training objective and final selection.")
print("Reward is reported rather than asserted as an improvement: the safety stage may legitimately trade some return for risk reduction, while the checkpoint selector minimizes unnecessary return erosion.")
```

## A methodological note worth stating plainly — this is the fix, not a leftover caveat

Unlike the original build of this stage, this training run happens on **TRAIN** (2018-2024), not
on the TEST period the final comparison will be scored on. That was the whole point of the
retraining exercise: `Step6B_WalkForwardEvaluation.ipynb` applies this policy, once, at the very
end, to `bl_llm_memory_weights.csv` — the already-optimized TEST-range weights this policy has
never seen in any form during the training above. The numbers in this notebook (reward, cost,
violation counts) are still an honest description of how the *training* run went, on TRAIN — they
are not, and were never meant to be, the final performance claim. That claim is what
`Step6B_WalkForwardEvaluation.ipynb` produces, genuinely out-of-sample this time.
